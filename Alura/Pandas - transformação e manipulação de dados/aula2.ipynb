Dados numéricos

```python
import pandas as pd
import numpy as np
```

```python
path = 'dados/dados_hospedagem.json'

dados = pd.read_json(path)

dados.head(5)
```

```python
dados = pd.json_normalize(dados['info_moveis'])
```

```python
dados.head(3)
```

Da coluna 3 em diante, percebe que os dados estão em lista?

```python
dados.iloc[1,3]
```

Neste exemplo, é interessante explodir essa lista, para que visualizemos todo o conteúdo dela

```python
colunas = list(dados.columns)
colunas[:4]
```

```python
dados = dados.explode(colunas[3:])
dados.head(3)
```

Perceba que o index das linhas está bagunçado, afinal o que era uma linha só, transformou-se em várias linhas que herdaram o index da linha "mãe"

```python
dados.reset_index(inplace=True, drop=True)
dados.head()
```

Show, agora precisamos ver a tipagem das colunas. veja:

```python
dados.info()
```

max_hospedes, que é uma coluna numérica, está formatada como texto. precisamos concertar todos os casos assim.

```python
dados['max_hospedes'] = dados['max_hospedes'].astype('int64')
```

```python
colunas_numericas = ['quantidade_banheiros', 'quantidade_camas', 'quantidade_quartos']

dados[colunas_numericas] = dados[colunas_numericas].astype('int64')
dados['avaliacao_geral'] = dados['avaliacao_geral'].astype('float64')

dados.info()
```

Você vai perceber que as vezes é necessário tratar colunas de numeros que contenham letra. veja o caso da coluna preco

```python
dados['preco'].head()
```

É necessário retirar o "$" e as vírgulas que podem aparecer.

```python
dados['preco'] = dados['preco'].apply(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
dados['preco'].head()
```

```python
dados['preco'] = dados['preco'].astype('float64')
```
