Conhecendo os dados

```python
import pandas as pd
```

```python
path = 'dados/1-SEEG10_GERAL-BR_UF_2022.10.27-FINAL-SITE.xlsx'
dados = pd.read_excel(path, sheet_name='GEE Estados')
```

```python
df = pd.DataFrame(dados)
df.info()
```

Selecionando valores únicos com unique

```python
df['Emissão / Remoção / Bunker'].unique()
```

Filtrando dataframes via comparação booleana

```python
df[(df['Emissão / Remoção / Bunker'] == 'Remoção') | (df['Emissão / Remoção / Bunker'] == 'Remoção NCI')]
```

Usando isin

```python
remocoes = ['Remoção NCI', 'Remoção']
```

```python
df[df['Emissão / Remoção / Bunker'].isin(remocoes)]
```

Vimos que faz sentido. Mas tem um ponto: faz sentido além da amostra de dados?

Abaixo, selecionamos apenas o que faz sentido observar para amostra, usando a função loc, que é um método de filtragem mais avançado

```python
df.loc[df['Emissão / Remoção / Bunker'].isin(remocoes), 1970:2021]
```

```python
df.loc[df['Emissão / Remoção / Bunker'].isin(remocoes), 1970:2021].max()
```

```python
df.loc[df['Emissão / Remoção / Bunker'] == 'Bunker', 'Estado'].unique()
```
